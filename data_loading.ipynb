```python
import zstandard as zstd 
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import chess.pgn
import io 
```

```python
decompressor = zstd.ZstdDecompressor()
```

```python
with open("lichess_games.zst", "rb") as f:
    reader = decompressor.stream_reader(f)
    text_stream = io.TextIOWrapper(reader, encoding = "utf-8")

    games = []
    while True:
        game = chess.pgn.read_game(text_stream)
        if game is None: 
            break

        headers = game.headers
        if (headers.get("TimeControl") == "180+0"
            and int(headers.get("WhiteElo", 0)) > 2000):

            games.append({
                "white": headers["White"],
                "black": headers["Black"],
                "white_elo": headers["WhiteElo"],
                "black_elo": headers["BlackElo"],
                "result": headers["Result"],
                "opening": headers.get("Opening", ""),
                "moves": str(game.mainline_moves()),
            })
        if len(games) > 0 and len(games) % 1000 == 0:
            print(f"Finished {len(games)} games so far")
        if len(games) >= 50000:
            break
```

```python
df_subset = pd.DataFrame(games)
```
